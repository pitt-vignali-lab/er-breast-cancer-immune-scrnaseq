```python
# loading pacakges and settings
import anndata
import scvelo as scv
import pandas as pd
import scanpy as sc
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['savefig.facecolor']='#ffffff'
plt.rcParams['figure.facecolor']='#ffffff'
```

```python
# readin the loom file and check cell IDs
sample_one = anndata.read_loom("../../1.Data/SayaliVelocyto/combined.loom")
# sample_one = anndata.concat([sample1, sample2], join="outer")
sample_one.obs.head
```

```python
sample_one.obs=sample_one.obs.rename(index = lambda x: x.replace('_GEX', '').replace(':', '_').replace('x', ''))
sample_one.obs.head
```

```python
# Load information from Seurat project
sample_obs = pd.read_csv("cellID.csv")
sample_obs
```

```python
umap = pd.read_csv("umap.csv")
umap = umap.rename(columns = {"Unnamed: 0":'CellID'})
cell_clusters = pd.read_csv("clusters.csv")
cell_clusters = cell_clusters.rename(columns = {"Unnamed: 0":'CellID'})
cell_types = pd.read_csv("types.csv")
cell_types = cell_types.rename(columns = {"Unnamed: 0":'CellID'})
cell_types.head
```

```python
# Combine seurat and volecity 
sample_one = sample_one[np.isin(sample_one.obs.index,sample_obs["x"])]
sample_one.obs
```

```python
sample_one_index = pd.DataFrame(sample_one.obs.index)
sample_one_index = sample_one_index.rename(columns = {0:'CellID'})

umap_ordered = sample_one_index.merge(umap,on="CellID")
cell_clusters = sample_one_index.merge(cell_clusters,on="CellID")
cell_types = sample_one_index.merge(cell_types, on = "CellID")

umap_ordered = umap_ordered.iloc[:,1:]
sample_one.obsm['X_umap'] = umap_ordered.values

cell_clusters_ordered=cell_clusters.iloc[:,1]
sample_one.obs['cell_clusters']=cell_clusters_ordered.values

cell_types_ordered =  cell_types.iloc[:,1:]
sample_one.obs['cell_types'] = cell_types_ordered.values
```

```python
#Running RNA Velocity and ploting
scv.pp.filter_and_normalize(sample_one,
                            min_shared_counts=20,
                            n_top_genes=2000)
scv.pp.moments(sample_one, n_pcs=30, n_neighbors=30)
scv.tl.velocity(sample_one, mode = "stochastic")
scv.tl.velocity_graph(sample_one)
```
